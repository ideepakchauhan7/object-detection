# YOLOv8 Inference Demo

Run inference from a notebook with the same config used by the scripts.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if str(PROJECT_ROOT / 'src') not in sys.path:
    sys.path.append(str(PROJECT_ROOT / 'src'))

from utils import load_config, resolve_weights
from ultralytics import YOLO

config = load_config('config.yaml')
weights = resolve_weights(None, config)
model = YOLO(weights)
weights
```

```python
# Replace this with a local image path, or use a URL supported by Ultralytics.
source = 'https://ultralytics.com/images/bus.jpg'
results = model.predict(source=source, conf=0.25, save=True)
results[0].show()
```
